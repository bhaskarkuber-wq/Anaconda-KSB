```python
#5.	Meteor Strikes
#A region experiences meteor strikes at an average rate of 1 per year.
#  What is the probability that there will be no meteor strikes in a given year?
#  What is the probability that there will be at least 2 meteor strikes?

from scipy.stats import poisson

lamda_value=1

#X=0
x_zero=poisson.pmf(0,lamda_value)

print(f"probability that there will be no meteor strikes in a given year:{x_zero}")

#X>=2 i.e. 1-P(X<2) = 1 - ( P(X=0) + P(X=1) )
x_one=poisson.pmf(1,lamda_value)

print(f"probability that there will be at least 2 meteor strikes:{1-(x_zero+x_one)}")





```
