```python
#5. Lambda and recursion 
#Use a lambda expression to sort the students in descending order of average marks.
def sort_students():
    sorted_students = sorted(students, key=lambda student: student["average"],reverse=True)
    return(sorted_students)

#Create a recursive function find_max(values) that accepts a list of numeric values and returns the maximum value. 
#The function must have a clear base case and recursive case.
def find_max_recursive(numbers):
    
    if len(numbers) == 1:
        return numbers[0]
    
    # Recursive case: compare first element with max of the rest
    max_of_rest = find_max_recursive(numbers[1:])
    return numbers[0] if numbers[0] > max_of_rest else max_of_rest


# Example usage

nums = [12, 145, 23, 67, 34, 89, 21]

print(f"\n\nFinding max value from the given list {nums}: {find_max_recursive(nums)}")

print(f"\n\nBefore sorting:-->")
display_student(students)

#sorting the student list based on average marks
sorted_students=sort_students()
print(f"\n\nAfter sorting (based on descending order of average marks):-->")
display_student(sorted_students)
    
```
