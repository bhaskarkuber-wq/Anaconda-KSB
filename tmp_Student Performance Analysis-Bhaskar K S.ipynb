```python
counter=0

def find_max_recursive(numbers):
    # Base case: if the list has only one element, return it
    global counter
    if len(numbers) == 1:
        #counter+=1        
        #print(f"count:{counter} ,numbers[0] ->{numbers[0]}")
        return numbers[0]
    
    # Recursive case: compare first element with max of the rest
    max_of_rest = find_max_recursive(numbers[1:])
    #print("max_of_rest -?",max_of_rest)
    #print(f"number:",numbers)
    return numbers[0] if numbers[0] > max_of_rest else max_of_rest


# Example usage
nums = [12, 145, 23, 67, 34, 89, 21]
print("Maximum value:", find_max_recursive(nums))


```
