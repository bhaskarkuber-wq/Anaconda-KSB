```python
#random number generation
import numpy as np
from numpy import random

rnd=random.default_rng(101) # see value specifed to generate same set of random vectors

rnd.integers(1,6,size=5,endpoint=True)
```

```python
l_float=rnd.random(50)
print(l_float)
```

```python
import matplotlib.pyplot as plt
x=rnd.integers(1,50,size=100)
y=rnd.integers(1,40,size=100)

print(f"\nx: {np.sort(x)}")
print(f"\ny: {y}")
plt.scatter(x,y)
plt.show
```

```python
rnd.shuffle(x)
rnd.shuffle(y)
plt.scatter(x,y)
plt.show
print(f"\nx: {np.sort(x)}")
print(f"\ny: {y}")
```

```python
#choice(0 this fucntion can randonmly select number of vector elements
x_selected = rnd.choice(x,size=60,replace=False)
y_selected = rnd.choice(y,size=60)
print(f"\nx_selected: {x_selected}")
print(f"\ny_selected: {y_selected}")
```
