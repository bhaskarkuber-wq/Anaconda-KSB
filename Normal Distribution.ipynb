```python
from statsmodels.stats.weightstats import ztest
from scipy import stats

import numpy as np

fill_volume = np.array([
    503, 505, 499, 504, 506, 502, 501, 507, 500, 504,
    503, 506, 502, 505, 504, 501, 508, 503, 505, 502,
    504, 506, 503, 501, 507, 504, 502, 505, 503, 506
])

print("n =", len(fill_volume))
print("Sample mean =", fill_volume.mean())

#type 1 mu=500 h1 mu!=500
#type 1 two sided 

#type1
z_stat, p_value=ztest(fill_volume,value=500, alternative="two-sided")
print(f"z statistics={z_stat:.4f}")
print(f"p-value={p_value:.4f}")
print()

#type 1 mu=500 h1 mu!=500
#type 1 two-sided type2 larger type3 smaller

#type2
z_stat, p_value=ztest(fill_volume,value=500, alternative="larger")
print(f"z statistics={z_stat:.4f}")
print(f"p-value={p_value:.4f}")
print()

#type3
z_stat, p_value=ztest(fill_volume,value=500, alternative="smaller")
print(f"z statistics={z_stat:.4f}")
print(f"p-value={p_value:.4f}")
```

```python

from scipy import stats

import numpy as np

fill_volume = np.array([
    503, 505, 499, 504, 506, 502, 501, 507, 500, 504,
    503, 506, 502, 505, 504, 501, 508, 503, 505, 502,
    504, 506, 503, 501, 507, 504, 502, 505, 503, 506
])

print("n =", len(fill_volume))
print("Sample mean =", fill_volume.mean())


#type 1 mu=500 h1 mu!=500
#type 1 two sided 

#type1
t_stat, p_value=stats.ttest_1samp(fill_volume,popmean=500, alternative="two-sided")
print(f"t statistics={t_stat:.4f}")
print(f"p-value={p_value:.4f}")
print()

#type 1 mu=500 h1 mu!=500
#type 1 two-sided type2 greater type3 less

#type2
t_stat, p_value=stats.ttest_1samp(fill_volume,popmean=500, alternative="greater")
print(f"t statistics={t_stat:.4f}")
print(f"p-value={p_value:.4f}")
print()

#type3
t_stat, p_value=stats.ttest_1samp(fill_volume,popmean=500, alternative="less")
print(f"t statistics={t_stat:.4f}")
print(f"p-value={p_value:.4f}")

```

```python
from statsmodels.stats.weightstats import ztest
from scipy import stats

import numpy as np

c=[2,-1,0,3,5]
cbar=np.mean(c)
s=np.std(c,ddof=1)
n=5

t_test=(cbar-0)/(s/pow(5,.5))
print("Probablitiy of density fucntion:",t_test)

pro=stats.t.sf(t_test,n-1)
print("Probablitiy of density fucntion::",pro)

#type3
t_stat, p_value=stats.ttest_1samp(c,popmean=0, alternative="greater")
print(f"t statistics={t_stat:.4f}")
print(f"p-value={p_value:.4f}\n")

after=[15,12,10,9,20]
before=[13,13,10,6,15]
t_stat, p_value=stats.ttest_rel(before,after, alternative="less")

print(f"t statistics={t_stat:.4f}")
print(f"p-value={p_value:.4f}")
```

```python
import numpy as np
from scipy import stats
import pandas as pd


observed = np.array([100, 52, 48, 200])
expected = np.array([120, 160, 80, 40])

result = stats.chisquare(f_obs=observed, f_exp=expected)
print(result)

display(pd.DataFrame({
    "Defect type": ["A", "B", "C", "D"],
    "Observed": observed,
    "Expected": expected
}))
print(f"Chi-square statistic = {result.statistic:.4f}")
print(f"p-value = {result.pvalue:.4f}")
#print(pvalue_decision(result.pvalue))

#pvalue <5% -> h1 true which means the calim from the manager is incorrect
```

```python
contingency_table = pd.DataFrame(
    [[18, 82], [30, 70], [12, 88]],
    index=["Morning", "Evening", "Night"],
    columns=["Defective", "Not defective"]
)
display(contingency_table)

chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Chi-square statistic = {chi2:.4f}")
print(f"Degrees of freedom = {dof}")
print(f"p-value = {p_value:.4f}")
print(p_value>0.05)

#pvalue is less than .05, hence H1 is accepted, H0 rejected.abs

```

```python
pg1 = np.array([2,2.5,2.7,2,1.9])
pg2 = np.array([3.1,2.9,2.5,3,3.2])
pg3 = np.array([1.6,.5,1.7,1,.8])

m1=pg1.mean()
print(m1)
m2=pg2.mean()
print(m2)
m3=pg3.mean()
print(m3)

sum1=0
for i in range(pg1.size):
    sum1=sum1+((pg1[i]-m1)**2)

print(sum1)

sum2=0
for i in range(pg2.size):
    sum2=sum2+((pg2[i]-m2)**2)

print(sum2)
sum3=0
for i in range(pg3.size):
    sum3=sum3+((pg3[i]-m3)**2)

print(sum3)

dwg=(sum1+sum2+sum3)/(12)
print("dwg:",dwg)

grand_mean=(2+2.5+2.7+2+1.9++3.1+2.9+2.5+3+3.2+1.6+.5+1.7+1+.8)/15

print("Grand mean:",grand_mean)

dbg=(5*(m1-grand_mean)**2 + 5*(m2-grand_mean)**2 +5*(m3-grand_mean)**2 )/2

print("dbg:",dbg)


test_stats=(dbg/dwg)
print("test statistics:",test_stats)

print("\n")


print(stats.f_oneway(pg1,pg2,pg3))
```
