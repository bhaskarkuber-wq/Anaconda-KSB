# Box Plot

```python
v1 = list(range(200,300))
v1.append(10)
v1.append(20)
v1.append(400)
plt.boxplot(v1)
q1=np.percentile(v1,25)
q2=np.percentile(v1,50)
q3=np.percentile(v1,75)
iqr=q3-q1
low_whisker_limit=q1-(iqr*1.5)
higher_whisker_limit=q3+(iqr*1.5)
print(low_whisker_limit,higher_whisker_limit)
plt.axhline(y=low_whisker_limit)
plt.axhline(y=higher_whisker_limit)
plt.show()
```

```python

plt.boxplot(v1,orientation="horizontal")
q1=np.percentile(v1,25)
q2=np.percentile(v1,50)
q3=np.percentile(v1,75)
iqr=q3-q1
low_whisker_limit=q1-(iqr*1.5)
higher_whisker_limit=q3+(iqr*1.5)
print(low_whisker_limit,higher_whisker_limit)
plt.axvline(x=low_whisker_limit)
plt.axvline(x=higher_whisker_limit)
plt.show()
```
