```python
#vector project

g=np.array([1,2])
h=np.array([2,3])
print(g)
print(h)
g_projection_h=(np.dot(g,h) / np.linalg.norm(h)) * (h/np.linalg.norm(h)) # to find magnitude of matrix h np.linalg.norm(b) 
h_projection_g=(np.dot(g,h) / np.linalg.norm(g)) * (g/np.linalg.norm(g))# to find magnitude of matrix g np.linalg.norm(b) 

print("g vector projecion over h",g_projection_h) #scalar project 
print("h vector projecion over g",h_projection_g)
```

```python
v1=np.array([1,2,3])
v2=np.array([7,9,1])
v3=np.array([2,4,0])

def projection_a_over_b(g,h):
    return (np.dot(g,h) / np.linalg.norm(h)) * (h/np.linalg.norm(h))

x1=v1
#x2=v2-pro(v2 on v1)

x2=v2-(projection_a_over_b(v2,v1))

print(x1)
print(x2)

print(np.dot(x1,x2))
```

```python
x3=v3-(projection_a_over_b(v3,x1))-projection_a_over_b(v3,x2)
print(x3)
```

```python
np.dot(x1,x2), np.dot(x1,x3), np.dot(x2,x3)
```
